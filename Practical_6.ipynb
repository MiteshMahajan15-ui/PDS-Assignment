## Theory

### 1. What is web scraping? Explain the role of the Requests library in web scraping.

**Web Scraping:** Web scraping is the automated process of extracting data from websites. It involves programmatically fetching web pages and parsing their content to collect specific information, which can then be stored, analyzed, or used for various purposes.

**Role of Requests Library:** The Requests library in Python is crucial for web scraping as it simplifies sending HTTP requests (like GET and POST) to web servers. It allows you to fetch the raw HTML content of a web page, which is the first step in web scraping. Without a library like Requests, you wouldn't be able to retrieve the web page's source code to begin parsing.

### 2. What is BeautifulSoup? How does it help in parsing HTML?

**BeautifulSoup:** BeautifulSoup is a Python library designed for parsing HTML and XML documents. It creates a parse tree from page source code that can be used to extract data in a hierarchical and readable way.

**How it helps in parsing HTML:** After fetching a web page's HTML content (e.g., using Requests), BeautifulSoup takes this raw HTML and transforms it into a navigable tree of Python objects. This tree structure allows you to:
*   **Navigate:** Easily traverse the HTML document by tag names, IDs, classes, etc.
*   **Search:** Find specific elements using various methods like find(), find_all(), select(), etc.
*   **Extract:** Retrieve text, attributes, or other content from the selected elements.

This makes it much easier to pinpoint and extract the desired data from complex HTML structures.

### 3. Differentiate between web scraping and web crawling.

**Web Scraping:** Focuses on extracting specific data from a web page or a set of web pages. It's about what data is taken from a page.
*   **Example:** Collecting product prices and names from an e-commerce site.

**Web Crawling:** Focuses on discovering and indexing web pages by following hyperlinks from one page to another. It's about exploring the web.
*   **Example:** A search engine bot visiting links to build its index of the internet.

**Relationship:** Web crawling often precedes web scraping. A crawler might find pages that a scraper then processes to extract data.

### 4. What are the commonly used methods in BeautifulSoup?

Some commonly used methods in BeautifulSoup include:
*   BeautifulSoup(html_doc, 'html.parser'):- The constructor to create a BeautifulSoup object, parsing the HTML document.
*   find(name, attrs, recursive, string, kwargs):- Finds the first tag that matches the given criteria.
*   find_all(name, attrs, recursive, string, limit, kwargs) (or findAll):- Finds all tags that match the given criteria.
*   select(selector):- Allows searching for elements using CSS selectors, which can be very powerful.
*   get_text():- Extracts all the text within a tag, including its descendants.
*   .name:- Returns the tag name.
*   .attrs:- Returns a dictionary of a tag's attributes.
*   .string:- Returns the NavigableString content of a tag (if it's a single string).
*   **Accessing attributes:**- tag['_attribute_name_'] (e.g., link['_href_'])

### 5. What are the ethical and legal considerations in web scraping?

Web scraping involves several ethical and legal considerations:
*   **Terms of Service (ToS):-** Many websites' ToS explicitly prohibit automated scraping. Violating ToS could lead to legal action, account termination, or IP blocking.
*   **Copyright:-** Scraped data might be copyrighted. Using or republishing copyrighted content without permission can lead to infringement claims.
*   **Data Privacy (GDPR, CCPA, etc.):-** If scraping personal data, privacy regulations must be strictly followed. Misuse of personal data can result in significant fines and legal penalties.
*   **Server Load:-** Aggressive scraping can overload a website's servers, leading to denial of service for legitimate users. This is unethical and can be illegal (e.g., considered a DDoS attack).
*   robots.txt: Websites often have a robots.txt file that specifies which parts of the site crawlers (and by extension, scrapers) are allowed or disallowed to access. Respecting robots.txt is an ethical best practice, though not always legally binding.
*   **Commercial Use:** The legality and ethical implications often differ significantly for personal, research, or commercial use of scraped data.
*   **Transparency:** Being transparent about scraping activities (e.g., identifying your scraper via a User-Agent header) can sometimes mitigate issues.

```python
import requests
from bs4 import BeautifulSoup

# Step 1: Define the URL of the webpage to scrape
url = "https://example.com"

# Step 2: Define headers to simulate a real browser request
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"
}

try:
    # Step 3: Send HTTP request to the server
    response = requests.get(url, headers=headers, timeout=10)

    # Step 4: Check if request is successful
    if response.status_code == 200:
        print("Webpage fetched successfully!\n")
    else:
        print("Failed to fetch webpage. Status Code:", response.status_code)

    # Step 5: Parse the HTML content using BeautifulSoup
    soup = BeautifulSoup(response.text, "html.parser")

    # Step 6: Extract and display the title of the webpage
    if soup.title:
        print("Page Title:")
        print(soup.title.string, "\n")
    else:
        print("No title found\n")

    # Step 7: Extract all links (<a> tags)
    print("List of Links:\n")
    links = soup.find_all("a")

    link_count = 0
    for link in links:
        href = link.get("href")
        text = link.get_text(strip=True)

        if href:
            link_count += 1
            print(f"{link_count}. Text: {text if text else 'No text'}")
            print(f"   URL: {href}\n")

    # Step 8: Extract headings (h1, h2, h3)
    print("Headings:\n")

    for tag in ["h1", "h2", "h3"]:
        headings = soup.find_all(tag)

        if headings:
            print(f"--- {tag.upper()} Tags ---")
            for heading in headings:
                print(heading.get_text(strip=True))
            print()

    # Step 9: Extract paragraph data
    print("Paragraphs:\n")
    paragraphs = soup.find_all("p")

    para_count = 0
    for para in paragraphs:
        text = para.get_text(strip=True)
        if text:
            para_count += 1
            print(f"{para_count}. {text}\n")

    # Step 10: Extract image sources
    print("Image Sources:\n")
    images = soup.find_all("img")

    img_count = 0
    for img in images:
        src = img.get("src")
        if src:
            img_count += 1
            print(f"{img_count}. {src}")

    # Step 11: Display summary of extracted data
    print("\nSummary:")
    print("Total Links:", link_count)
    print("Total Headings:",
          len(soup.find_all('h1')) +
          len(soup.find_all('h2')) +
          len(soup.find_all('h3')))
    print("Total Paragraphs:", para_count)
    print("Total Images:", img_count)

# Step 12: Handle errors properly
except requests.exceptions.RequestException as e:
    print("Error in HTTP request:", e)

except Exception as e:
    print("Unexpected error occurred:", e)
```
