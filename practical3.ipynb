**1. What is data wrangling? Explain the difference between data wrangling and data preprocessing.**

*   **Data Wrangling:**
    *   Also known as data munging, is the process of cleaning and preparing raw data for analysis.
    *   This includes tasks such as removing or correcting erroneous data, converting data into the appropriate format, handling missing values, and reshaping or restructuring data to suit analysis needs.
    *   The goal is to ensure that the data is accurate, complete, and structured correctly for analysis.

*   **Difference Between Data Wrangling and Data Preprocessing:**
    *   **Data Wrangling:** Refers to the overall process of transforming raw data into a more usable and structured form. It includes various tasks like cleaning, transforming, and reshaping the data.
    *   **Data Preprocessing:** Is more specific and often refers to the steps taken before feeding the data into a machine learning model. It includes tasks like normalization, standardization, encoding categorical variables, and handling missing values.
    *   In short, data wrangling is broader and more general, while data preprocessing is typically a subset of wrangling specifically aimed at preparing data for modeling.

**2. What is the importance of exploratory data analysis (EDA) in Pandas?**

*   **Exploratory Data Analysis (EDA):**
    *   Is the initial phase of data analysis where we use statistical graphics, plots, and other tools to understand the structure, distribution, and patterns in the dataset.

*   **In the context of Pandas, EDA is critical because:**
    *   **Identifies Patterns and Relationships:** Helps uncover hidden patterns, correlations, and trends in the data, which can inform the choice of the model.
    *   **Detects Outliers or Anomalies:** Helps detect outliers, missing values, and errors in the dataset that need to be addressed before further analysis.
    *   **Informs Feature Engineering:** By understanding the relationships between variables, EDA can guide the creation of new features or the transformation of existing features.
    *   **Improves Understanding of Data:** Provides a deep understanding of the dataset’s behavior, distribution, and underlying structure, which is crucial for making informed decisions about the data.
    *   In Pandas, EDA can be performed through methods like `.describe()`, `.info()`, `.value_counts()`, `.corr()`, and using visualization libraries like matplotlib and seaborn.

**3. What are the main differences between Series, DataFrame, and Panel?**

*   **Series:**
    *   A one-dimensional labeled array capable of holding any data type (integers, strings, floats, etc.).
    *   Essentially a single column of data.
    *   Example: A list of numbers or strings with an index.
    *   Syntax: `pd.Series(data, index=labels)`

*   **DataFrame:**
    *   A two-dimensional labeled data structure with columns of potentially different types.
    *   It can be thought of as a collection of Series objects that share a common index.
    *   Example: A table or a spreadsheet.
    *   Syntax: `pd.DataFrame(data, columns=labels)`

*   **Panel (Deprecated in Pandas 1.0+):**
    *   A three-dimensional data structure. Think of it as a DataFrame with an additional axis.
    *   It allows for holding data in three dimensions: rows, columns, and a third dimension (e.g., time, items, etc.).
    *   Example: Time-series data with multiple attributes per time period.
    *   Syntax: `pd.Panel(data)` (Note: Panel is deprecated in favor of xarray for multi-dimensional data).

**4. Explain the use of:**

*   **a. `isnull()`:**
    *   The `isnull()` function is used to detect missing values in a DataFrame or Series.
    *   It returns a boolean DataFrame/Series where `True` represents missing values (NaN) and `False` represents non-missing values.
    *   Example: `df.isnull()`

*   **b. `notnull()`:**
    *   The `notnull()` function is the inverse of `isnull()`.
    *   It returns `True` for non-missing values and `False` for missing values.
    *   Example: `df.notnull()`

*   **c. `dropna()`:**
    *   The `dropna()` function is used to remove any rows or columns that contain missing values (NaN).
    *   You can specify whether to drop rows or columns, and how many missing values must be present to drop the entry.
    *   Example: `df.dropna()`  # Drops rows with any NaN values

*   **d. `fillna()`:**
    *   The `fillna()` function is used to replace missing values with a specific value (such as the mean, median, or a custom value).
    *   It can also perform forward-fill or backward-fill to propagate the previous or next non-null value.
    *   Example: `df.fillna(0)`  # Replaces NaN with 0
    *   Example: `df.fillna(df.mean())`  # Replaces NaN with column-wise mean

**5. What is column transformation and why is it important in data analysis?**

*   **Column Transformation:**
    *   Column transformation refers to the process of modifying one or more columns in a dataset to improve the quality or relevance of the data for analysis or modeling.
    *   This can involve several techniques such as scaling, normalization, encoding, or imputation.

*   **Types of Column Transformations:**
    *   **Scaling:** Adjusting the range of data (e.g., Min-Max Scaling, Standardization).
    *   **Encoding:** Converting categorical variables into numerical values (e.g., One-Hot Encoding, Label Encoding).
    *   **Imputation:** Filling in missing values (e.g., replacing NaN with the mean, median, or mode).
    *   **Log Transformation:** Applying a log function to transform skewed distributions into more normal distributions.

*   **Importance:**
    *   **Improves Model Performance:** Column transformations like scaling or normalization make the data more consistent, helping many machine learning algorithms perform better.
    *   **Handles Skewness and Outliers:** Techniques like log transformations or handling outliers ensure that the data doesn't distort the results.
    *   **Improves Interpretability:** Transforming categorical variables into numerical forms allows machine learning algorithms to process them efficiently.
    *   **Prepares Data for Algorithms:** Some algorithms (e.g., SVM, K-means) are sensitive to the scale of the data, and column transformations ensure that data is suitable for these models.


```python
import pandas as pd
import numpy as np

df = pd.read_csv("/content/cleaned_employee_data.csv")

print(df[df.columns[0]])

df["Status"] = "Active"

df_sorted = df.sort_values(by=df.columns[0])

df = df.rename(columns={df.columns[0]: "ID"})

print(df.describe(include="all"))

if len(df.select_dtypes(include=np.number).columns) >= 2:
    c1, c2 = df.select_dtypes(include=np.number).columns[:2]
    filtered = df[(df[c1] > df[c1].mean()) & (df[c2] > df[c2].mean())]
    print(filtered)

cat_cols = df.select_dtypes(include="object").columns
num_cols = df.select_dtypes(include=np.number).columns

if len(cat_cols) > 0 and len(num_cols) > 0:
    grouped = df.groupby(cat_cols[0])[num_cols[0]].mean()
    print(grouped)

df2 = pd.DataFrame({"ID": df["ID"], "Bonus": range(len(df))})

merged = pd.merge(df, df2, on="ID")
print(merged.head())

df_drop = df.drop(index=df.index[:2], errors="ignore")
df_drop = df_drop.drop(columns=["Status"], errors="ignore")

if len(num_cols) > 0:
    df[num_cols[0]] = df[num_cols[0]].apply(lambda x: x * 2)

if len(cat_cols) > 0 and len(num_cols) > 0:
    pivot = df.pivot_table(values=num_cols[0], index=cat_cols[0], aggfunc="mean")
    print(pivot)

    melted = pd.melt(df, id_vars=["ID"])
    print(melted.head())

if len(cat_cols) >= 1:
    multi = df.set_index([cat_cols[0], "ID"])
    print(multi.head())

join_df = df.set_index("ID").join(df2.set_index("ID"))
print(join_df.head())

if len(cat_cols) > 0 and len(num_cols) > 0:
    custom = df.groupby(cat_cols[0])[num_cols[0]].agg(lambda x: x.max() - x.min())
    print(custom)

df["Date"] = pd.date_range("2024-01-01", periods=len(df), freq="D")

ts = df.set_index("Date").resample("ME").size()
print(ts)
```
