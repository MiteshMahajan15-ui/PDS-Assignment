a) Missing values in a dataset can be identified using functions provided by the Pandas library.
b) The isnull() function is used to check whether a value is missing in the dataset.

c) When combined with sum(), it counts the number of missing values in each column.

```python
import pandas as pd
import numpy as np

# Create a dummy CSV file for demonstration purposes if it doesn't exist
try:
    df = pd.read_csv("employee_performance_data.csv")
except FileNotFoundError:
    print("File 'employee_performance_data.csv' not found. Creating a dummy file.")
    data = {
        'Age': [25, 30, np.nan, 35, 28, 40, 25, 32, 29, np.nan],
        'Gender': ['Male', 'Female', 'Male', 'Female', 'Male', np.nan, 'Male', 'Female', 'Male', 'Female'],
        'Salary': [50000, 60000, 55000, np.nan, 70000, 80000, 50000, 62000, 65000, 58000],
        'PerformanceScore': [85, 92, 78, 88, np.nan, 95, 85, 90, 89, 82],
        'Department': ['Sales', 'Marketing', 'HR', 'Sales', 'IT', 'Marketing', 'HR', 'IT', 'Sales', 'Marketing']
    }
    dummy_df = pd.DataFrame(data)
    dummy_df.to_csv("employee_performance_data.csv", index=False)
    df = pd.read_csv("employee_performance_data.csv")

print("First 5 Rows:")
print(df.head())

print("\nMissing Values:")
print(df.isnull().sum())

df_zero = df.fillna(0)
print("\nDataset after replacing missing values with 0:")
print(df_zero.head())

df.columns = df.columns.str.lower()
print("\nColumn Names in Lowercase:")
print(df.columns)

print("\nDataset Shape:")
print(df.shape)

df_no_duplicates = df.drop_duplicates()
print("\nShape after removing duplicates:")
print(df_no_duplicates.shape)

for col in df.select_dtypes(include=np.number).columns:
    df[col] = df[col].fillna(df[col].mean())

for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].fillna(df[col].mode()[0])

print("\nMissing values handled using mean/mode.")

if 'age' in df.columns:
    df['age'] = df['age'].astype(int)

if 'salary' in df.columns:
    filtered_df = df[df['salary'] > 50000]
    print("\nFiltered Rows (salary > 50000):")
    print(filtered_df.head())

df.to_csv("cleaned_employee_data.csv", index=False)
print("\nCleaned dataset saved as cleaned_employee_data.csv")

df_zero_fill = df.fillna(0)
df_mean_fill = df.fillna(df.mean(numeric_only=True))
print("\nCompared Zero Fill vs Mean Fill")

if 'gender' in df.columns:
    df['gender'] = df['gender'].str.lower().replace({
        'm': 'male',
        'f': 'female'
    })

def clean_data(data):
    data.columns = data.columns.str.lower()
    data = data.drop_duplicates()

    for col in data.select_dtypes(include=np.number).columns:
        data[col] = data[col].fillna(data[col].mean())

    for col in data.select_dtypes(include='object').columns:
        data[col] = data[col].str.strip().str.lower()
        data[col] = data[col].fillna(data[col].mode()[0])

    return data

df_cleaned = clean_data(df)

for col in df_cleaned.select_dtypes(include=np.number).columns:
    Q1 = df_cleaned[col].quantile(0.25)
    Q3 = df_cleaned[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df_cleaned[(df_cleaned[col] < lower) | (df_cleaned[col] > upper)]

    print(f"\nOutliers in {col}: {len(outliers)}")

report = {
    "Original Rows": len(df),
    "Rows After Duplicate Removal": len(df_no_duplicates),
    "Columns": len(df.columns),
    "Missing Values Remaining": df_cleaned.isnull().sum().sum()
}

print("\nData Cleaning Report:")
for key, value in report.items():
    print(f"{key}: {value}")
```
