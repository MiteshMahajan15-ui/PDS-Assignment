1.  **Importance of data visualization in data analysis and its role in EDA:**
    *   **Importance:** Data visualization is crucial because it allows us to understand complex datasets quickly and intuitively. It transforms raw data into graphical representations (charts, graphs, maps), making patterns, trends, outliers, and relationships more apparent than they would be in tables of numbers. This aids in decision-making, hypothesis generation, and communication of insights.
    *   **Role in EDA (Exploratory Data Analysis):** In EDA, visualization is fundamental for:
        *   **Understanding Data Distribution:** Histograms and box plots help to see the spread, central tendency, and skewness of individual variables.
        *   **Identifying Relationships:** Scatter plots reveal correlations between two variables, while pair plots extend this to multiple variable pairs.
        *   **Detecting Outliers:** Visualizations make it easy to spot unusual data points that might indicate errors or significant events.
        *   **Pattern Recognition:** Line plots can show trends over time, and bar plots can compare categories, revealing underlying patterns.
        *   **Data Quality Assessment:** Visualizing data can highlight missing values or inconsistencies that need to be addressed during data cleaning.

2.  **Difference between Matplotlib and other visualization libraries:**
    *   **Matplotlib:** It is the foundational plotting library for Python. It provides a highly flexible, low-level interface for creating static, interactive, and animated visualizations in Python. It's often compared to MATLAB's plotting capabilities. Its strength lies in its fine-grained control over every element of a plot, making it suitable for creating highly customized and publication-quality figures. However, this flexibility can also mean it requires more code for common plots.
    *   **Other Libraries (e.g., Seaborn, Plotly, Bokeh):**
        *   **Seaborn:** Built on top of Matplotlib, Seaborn provides a higher-level interface for drawing attractive and informative statistical graphics. It simplifies the creation of complex plots (like heatmaps, violin plots, and regression plots) and handles common statistical tasks automatically. It's great for quickly exploring relationships in data and producing aesthetically pleasing plots with less code.
        *   **Plotly:** An interactive plotting library that creates web-based visualizations. It's excellent for creating interactive dashboards, scientific charts, and 3D plots that can be embedded in web applications. It's not limited to Python and has APIs for other languages.
        *   **Bokeh:** Another interactive visualization library that targets modern web browsers for presentation. It allows for highly customized interactive plots, dashboards, and data applications, often integrating with large or streaming datasets.
    *   **Key Differences:** Matplotlib offers maximum control but can be verbose. Libraries like Seaborn abstract away complexity for statistical plots, while Plotly and Bokeh focus on interactive, web-based visualizations.

3.  **Purpose of the `pyplot` module in Matplotlib:**
    *   `matplotlib.pyplot` is a collection of command-style functions that make Matplotlib work like MATLAB. Each `pyplot` function makes some change to a figure: e.g., creates a figure, creates a plotting area in a figure, plots some lines in a plotting area, decorates the plot with labels, etc.
    *   It's the most commonly used module for basic plotting tasks because it provides a simple interface to quickly create figures and axes, and then add elements to them. It automatically creates figures and axes as needed, simplifying the plotting workflow, especially for single plots.

4.  **Process of creating a basic line plot:**
    *   **Import `pyplot`:** Start by importing the `pyplot` module, typically aliased as `plt`:
        ```python
        import matplotlib.pyplot as plt
        ```
    *   **Prepare Data:** Define your data for the x-axis and y-axis. This could be lists, arrays, or pandas Series.
        
    *   **Create Plot:** Use the `plt.plot()` function, passing your x and y data. By default, this creates a line plot.
  
    *   **Add Labels and Title (Optional but Recommended):** Improve readability.
      
    *   **Display Plot:** Use `plt.show()` to display the plot.
  

5.  **Importance of customization in data visualization and available options in Matplotlib:**
    *   **Importance of Customization:** Customization is vital for:
        *   **Clarity and Readability:** Adjusting colors, fonts, labels, and plot styles can make a graph easier to understand.
        *   **Highlighting Key Insights:** Emphasizing specific data points or trends helps the viewer focus on the most important information.
        *   **Target Audience:** Tailoring the visualization to the audience's technical level and preferences.
        *   **Branding and Aesthetics:** Ensuring plots align with corporate or publication styles.
        *   **Addressing Specific Questions:** Sometimes, standard plots don't perfectly answer a question, and customization allows fine-tuning.
    *   **Customization Options in Matplotlib:** Matplotlib offers extensive customization at almost every level:
        *   **Figure and Axes:** Control figure size (`plt.figure(figsize=(width, height))`), background color, axis limits (`plt.xlim()`, `plt.ylim()`), and aspect ratio.
        *   **Plot Elements:**
            *   **Line Plots:** Line color (`color`), linestyle (`linestyle`), linewidth (`linewidth`), marker style (`marker`), marker size (`markersize`).
            *   **Scatter Plots:** Marker shape (`marker`), color (`color` or `c`), size (`s`), transparency (`alpha`).
            *   **Bar Plots:** Bar color, width, edge color.
        *   **Labels and Titles:** Set x-axis label (`plt.xlabel()`), y-axis label (`plt.ylabel()`), and plot title (`plt.title()`), with options for font size, color, and position.
        *   **Legends:** Add a legend (`plt.legend()`) to explain different plot elements, with options for location, font size, and frame.
        *   **Ticks and Grids:** Customize tick marks and labels (`plt.xticks()`, `plt.yticks()`), and add grid lines (`plt.grid()`) with various styles.
        *   **Text and Annotations:** Add arbitrary text (`plt.text()`) or annotations (`plt.annotate()`) at specific data points.
        *   **Colors and Colormaps:** Use specific colors (hex codes, RGB tuples, named colors) or colormaps for visualizing data values.
        *   **Subplots:** Create multiple plots within a single figure using `plt.subplot()` or `plt.subplots()` to compare different views of the data.
        *   **Styles:** Use `plt.style.use()` to apply predefined visual styles (e.g., 'ggplot', 'seaborn', 'dark_background').

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("/content/cleaned_employee_data.csv")

print(df.head())
print(df.info())
print(df.describe())

sns.set_theme(style="whitegrid")

plt.figure(figsize=(10,6))
sns.scatterplot(
    data=df,
    x="experience",
    y="salary",
    hue="department",
    size="salary",
    palette="Set2",
    sizes=(40,200)
)
plt.title("Experience vs Salary", fontsize=16)
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.show()

plt.figure(figsize=(12,6))
df_sorted = df.sort_values("employee_id")
plt.plot(
    df_sorted["employee_id"],
    df_sorted["salary"],
    marker="o",
    linewidth=2,
    color="green"
)
plt.title("Employee ID vs Salary", fontsize=16)
plt.xlabel("Employee ID")
plt.ylabel("Salary")
plt.xticks(rotation=45)
plt.show()

plt.figure(figsize=(10,6))
avg_salary = df.groupby("department")["salary"].mean().sort_values()
sns.barplot(
    x=avg_salary.index,
    y=avg_salary.values,
    hue=avg_salary.index,  # Add hue to resolve FutureWarning
    palette="viridis",
    legend=False # Set legend=False if hue is the same as x for bar plot
)
plt.title("Average Salary by Department", fontsize=16)
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.xticks(rotation=30)
plt.show()

plt.figure(figsize=(10,6))
sns.histplot(
    df["age"],
    bins=10,
    kde=True,
    color="skyblue"
)
plt.title("Age Distribution", fontsize=16)
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()

plt.figure(figsize=(10,6))
sns.boxplot(
    x="department",
    y="salary",
    data=df,
    palette="pastel"
)
plt.title("Department wise Salary Distribution", fontsize=16)
plt.xlabel("Department")
plt.ylabel("Salary")
plt.xticks(rotation=30)
plt.show()

sns.pairplot(
    df.select_dtypes(include="number"),
    diag_kind="kde",
    corner=True,
    plot_kws={"alpha":0.7}
)
plt.show()
````
