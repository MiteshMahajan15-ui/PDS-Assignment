### 1. Sequence Data Types in Python
*   Sequence data types are used to store multiple values in an ordered form.
*   They allow storing data in a single variable.
*   Elements are accessed using indexes and positions.
*   Main sequence types in Python:
    *   List – stores changeable items.
    *   Tuple – stores fixed items.
    *   String – stores characters or text.
    *   Range – stores sequence of numbers.

#### Indexing
*   Indexing is used to access a specific element.
*   First element starts at index 0.
*   Second element starts at index 1.
*   Negative indexing starts from the end.
*   Example: -1 means last element.

#### Slicing
*   Slicing is used to get a part of sequence data.
*   It returns multiple elements at once.
*   Syntax: `start : end : step`
*   Start index is included, end index is excluded.
*   Works in list, tuple, and string.

### 2. Difference Between List and Tuple in Python
*   List is mutable, so values can be added, removed, or changed.
*   Tuple is immutable, so values cannot be changed after creation.
*   List uses square brackets `[ ]`.
*   Tuple uses round brackets `( )`.
*   Lists are more flexible.
*   Tuples are faster in execution.

#### Why prefer tuple over list?
*   Uses less memory than lists.
*   Better for storing permanent data.
*   Safer because values cannot be modified.
*   Faster when reading data.
*   Can be used as dictionary keys.

#### Real-world uses:
*   Coordinates like latitude and longitude.
*   RGB color values.
*   Employee fixed records.
*   Days of week or months.

### 3. Python Arrays vs Lists
*   A list can store different data types together.
*   An array stores only same type elements.
*   Arrays require importing the `array` module.
*   Arrays are mainly used for numeric values.

#### Differences:
*   List supports mixed data such as integer, string, float.
*   Array supports only one type such as all integers.
*   Array uses less memory than list.
*   Array is faster when working with numbers.
*   Lists are easier and more commonly used.

#### When to use array module?
*   Large numeric datasets.
*   Memory efficient storage.
*   Mathematical calculations.
*   Sensor values.
*   Repeated same type numbers.

### 4. How Dictionaries Work Internally
*   Dictionary stores data in key-value pairs.
*   Each key is unique and linked with a value.
*   Python uses a hash table internally.
*   Key is converted into a hash value.
*   Hash value decides memory location of data.

#### Why lookups are faster?
*   Values are found directly using keys.
*   No need to search item one by one.
*   Searching is very fast compared to lists.
*   Lists and tuples require sequential searching.

#### Uses:
*   Student records with roll number.
*   Login username and password systems.
*   Product information.
*   Phone contact lists.
*   Fast searching applications.

### 5. Difference Between Sets and Lists
*   List is ordered and keeps insertion order.
*   List allows duplicate values.
*   Set is unordered collection.
*   Set does not allow duplicate values.
*   Sets are mainly used for unique data.

#### How sets handle duplicates?
*   Duplicate values are automatically removed.
*   Only one copy of repeated value is stored.
*   This makes data clean and unique.

#### Uses of sets:
*   Remove duplicate values from list.
*   Membership testing.
*   Fast searching.

```python
from array import array

# a) STRING
print("---- STRING ----")
s = "Python Programming"

print("Original String:", s)
print("Length:", len(s))
print("First Character:", s[0])
print("Last Character:", s[-1])
print("Slice (0:6):", s[0:6])
print("Lowercase:", s.lower())
print("Uppercase:", s.upper())
print("Replace:", s.replace("Python", "Java"))
print("Count of 'm':", s.count("m"))
print("Find 'Prog':", s.find("Prog"))
print("Concatenation:", s + " Language")
print()

# b) LIST
print("---- LIST ----")
lst = [10, 20, 30, 40]

print("Original List:", lst)
lst.append(50)
print("After Append:", lst)
lst.insert(1, 15)
print("After Insert:", lst)
lst.remove(30)
print("After Remove:", lst)
lst.pop()
print("After Pop:", lst)
lst.sort()
print("Sorted List:", lst)
lst.reverse()
print("Reversed List:", lst)
print("Index 2:", lst[2])
print()

# c) ARRAY
print("---- ARRAY ----")
arr = array('i', [1, 2, 3, 4, 5])

print("Original Array:", arr)
arr.append(6)
print("After Append:", arr)
arr.insert(2, 10)
print("After Insert:", arr)
arr.remove(3)
print("After Remove:", arr)
print("Count of 4:", arr.count(4))
print("Index of 10:", arr.index(10))
print()

# d) TUPLE
print("---- TUPLE ----")
tup = (100, 200, 300, 400)

print("Original Tuple:", tup)
print("Length:", len(tup))
print("Index 1:", tup[1])
print("Slice (1:3):", tup[1:3])
print("Count of 200:", tup.count(200))
print("Index of 300:", tup.index(300))
print()

# e) DICTIONARY
print("---- DICTIONARY ----")
d = {"name": "Alice", "age": 21, "city": "Nagpur"}

print("Original Dictionary:", d)
print("Keys:", d.keys())
print("Values:", d.values())
print("Items:", d.items())
print("Access Name:", d["name"])
d["age"] = 22
print("Updated Age:", d)
d["course"] = "Python"
print("After Adding Key:", d)
d.pop("city")
print("After Removing City:", d)
print()

# f) SETS
print("---- SETS ----")
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

print("Set1:", set1)
print("Set2:", set2)
print("Union:", set1.union(set2))
print("Intersection:", set1.intersection(set2))
print("Difference:", set1.difference(set2))
print("Symmetric Difference:", set1.symmetric_difference(set2))
set1.add(7)
print("After Add:", set1)
set1.remove(2)
print("After Remove:", set1)
print()

# g) RANGE
print("---- RANGE ----")
r = range(1, 11, 2)

print("Range Object:", r)
print("Converted to List:", list(r))
print("First Value:", r[0])
print("Last Value:", r[-1])
print("Length:", len(r))

for i in r:
    print(i, end=" ")
```
